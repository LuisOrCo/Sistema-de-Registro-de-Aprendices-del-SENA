# **Sistema de Registro de Aprendices SENA**

Proyecto en python para la gestion de aprendices SENA mediante la consola.

1. Registrar la siguiente información mediante una función.


*   Tipo Doc
*   Documento
*   Nombres
*   Apellidos
*   Ficha
*   Programa






```python


aprendices=[]

class Aprendiz:
  def __init__(self, tipoDoc, documento, nombres, apellidos, ficha, programa):
     self.tipoDoc = tipoDoc
     self.documento = documento
     self.nombres = nombres
     self.apellidos = apellidos
     self.ficha = ficha



def registroAprendiz():
    tipoDoc = input("Ingrese el tipo de documento del aprendiz: ")
    documento = input("Ingrese el documento del aprendiz: ")
    nombres = input("Ingrese los nombres del aprendiz: ")
    apellidos = input("Ingrese los apellidos del aprendiz: ")
    ficha = input("Ingrese la ficha del aprendiz: ")
    programa = input("Ingrese el programa del aprendiz: ")

    aprendizres = Aprendiz(tipoDoc, documento, nombres, apellidos, ficha, programa)

    aprendices.append(aprendizres)

def renderizarAprendices(aprendices):
      for aprendiz in aprendices:
        print("===============Aprendices Inscritos====================")
        print(f"Tipo de documento: {aprendiz.tipoDoc}")
        print(f"Documento: {aprendiz.documento}")
        print(f"Nombres: {aprendiz.nombres}")
        print(f"Apellidos: {aprendiz.apellidos}")
        print(f"Ficha: {aprendiz.ficha}")


while True:
  registroAprendiz()
  res = input("Desearía hacer un registro de un aprendiz ? (responda si o no): ")

  if res == "no":
        renderizarAprendices(aprendices)
        break





```
